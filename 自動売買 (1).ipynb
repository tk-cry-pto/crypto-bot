```python
# =========================================
# 1週目〜3週目 完全版
# bitFlyer / BTC/JPY / EMAクロス
# =========================================

!pip -q install pandas requests matplotlib numpy

import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time

# =========================================
# 1週目：方針確認
# =========================================
print("【1週目の方針】")
print("取引所：bitFlyer（Lightning）")
print("銘柄：BTC/JPY")
print("時間足：1時間足")
print("戦略：EMAクロス（EMA20 / EMA50）")
print("損失：1回の取引で資金の1%")
print("目標：月利3%")
print("やらないこと：")
print("・販売所を使わない")
print("・アルトコインを触らない")
print("・レバレッジを上げない")
print("・戦略を増やさない")
print("・途中でルールを変えない")


# =========================================
# bitFlyer BTC/JPY 約定履歴 大量取得 → CSV保存
# Colab用
# =========================================

!pip -q install pandas requests tqdm

import requests
import pandas as pd
import time
from tqdm import tqdm

# =========================================
# 1. 設定
# =========================================
PRODUCT_CODE = "BTC_JPY"
BASE_URL = "https://api.bitflyer.com/v1/executions"

COUNT_PER_REQUEST = 500      # 1回の取得件数
MAX_REQUESTS = 1000         # 取得回数。まずは200回くらいから
SLEEP_SEC = 0.5              # API負荷対策

RAW_CSV = "bitflyer_btcjpy_executions.csv"
OHLCV_1H_CSV = "bitflyer_btcjpy_1h.csv"

# =========================================
# 2. 約定履歴を取得する関数
# =========================================
def fetch_executions(product_code="BTC_JPY", count=500, before=None):
    params = {
        "product_code": product_code,
        "count": count
    }

    if before is not None:
        params["before"] = before

    response = requests.get(BASE_URL, params=params, timeout=10)

    # エラー確認
    if response.status_code != 200:
        raise Exception(f"HTTP Error: {response.status_code}, {response.text}")

    data = response.json()

    # bitFlyer側のエラー形式対策
    if isinstance(data, dict) and "error_message" in data:
        raise Exception(data)

    return data

# =========================================
# 3. 大量取得
# =========================================
all_rows = []
before_id = None

for _ in tqdm(range(MAX_REQUESTS)):
    try:
        data = fetch_executions(
            product_code=PRODUCT_CODE,
            count=COUNT_PER_REQUEST,
            before=before_id
        )

        if not data:
            print("データが空になったので終了します")
            break

        all_rows.extend(data)

        # 一番古いidより前を次に取得
        ids = [row["id"] for row in data]
        before_id = min(ids)

        time.sleep(SLEEP_SEC)

    except Exception as e:
        error_text = str(e)
        print("エラー発生:", error_text)

        if "Execution history is limited to the most recent 31 days" in error_text:
            print("bitFlyerの取得制限：直近31日分まで到達したため終了します")
            break

        print("10秒待って再試行します")
        time.sleep(10)

print("取得件数:", len(all_rows))

# =========================================
# 4. DataFrame化
# =========================================
df = pd.DataFrame(all_rows)

print("列名:", df.columns.tolist())
display(df.head())

# =========================================
# 5. 重複削除・時刻変換・並び替え
# =========================================
df = df.drop_duplicates(subset="id")

df["exec_date"] = pd.to_datetime(df["exec_date"], utc=True, format='ISO8601')
df = df.sort_values("exec_date").reset_index(drop=True)

print("最古:", df["exec_date"].min())
print("最新:", df["exec_date"].max())
print("件数:", len(df))

# =========================================
# 6. 約定履歴CSV保存
# =========================================
df.to_csv(RAW_CSV, index=False, encoding="utf-8-sig")
print("約定履歴CSV保存完了:", RAW_CSV)

# =========================================
# 7. 1時間足OHLCVに変換
# =========================================
# bitFlyer executionsには price, size が入っています
ohclv = df.set_index("exec_date").resample("1H").agg({
    "price": ["first", "max", "min", "last"],
    "size": "sum"
})

ohclv.columns = ["open", "high", "low", "close", "volume"]
ohclv = ohclv.dropna().reset_index()

# datetime列に名前変更
ohclv = ohclv.rename(columns={"exec_date": "datetime"})

display(ohclv.tail())

# =========================================
# 8. 1時間足CSV保存
# =========================================
ohclv.to_csv(OHLCV_1H_CSV, index=False, encoding="utf-8-sig")
print("1時間足CSV保存完了:", OHLCV_1H_CSV)

# =========================================
# 2週目：bitFlyerから価格取得
# =========================================
print("\n=========================================")
print("2週目：bitFlyerからBTC/JPY価格取得")
print("=========================================")

url = "https://api.bitflyer.com/v1/ticker"
params = {"product_code": "BTC_JPY"}

res = requests.get(url, params=params)
data = res.json()

print("現在価格:", data["ltp"])
print("買い気配:", data["best_bid"])
print("売り気配:", data["best_ask"])
print("出来高:", data["volume"])
print("時刻:", data["timestamp"])

# 10件だけ連続取得
rows = []

for _ in range(10):
    res = requests.get(url, params=params)
    d = res.json()
    rows.append({
        "timestamp": d["timestamp"],
        "ltp": d["ltp"],
        "best_bid": d["best_bid"],
        "best_ask": d["best_ask"],
        "volume": d["volume"]
    })
    time.sleep(1)

ticker_df = pd.DataFrame(rows)

print("\n--- 取得したtickerデータ ---")
display(ticker_df)

# 練習用EMA
ticker_df["ema3"] = ticker_df["ltp"].ewm(span=3, adjust=False).mean()
ticker_df["ema5"] = ticker_df["ltp"].ewm(span=5, adjust=False).mean()

plt.figure(figsize=(12, 5))
plt.plot(ticker_df["ltp"], label="Price")
plt.plot(ticker_df["ema3"], label="EMA3")
plt.plot(ticker_df["ema5"], label="EMA5")
plt.title("bitFlyer BTC/JPY Ticker (Practice EMA)")
plt.legend()
plt.grid(True)
plt.show()

# =========================================
# 3週目：EMAクロス戦略を作る
# =========================================
print("\n=========================================")
print("3週目：EMAクロス戦略を作る")
print("=========================================")

# 練習用価格データを作成
np.random.seed(42)

rows = 300
base_price = 10000000  # BTC/JPYイメージ

price_changes = np.random.normal(loc=3000, scale=50000, size=rows)
prices = base_price + np.cumsum(price_changes)

df = pd.DataFrame({
    "close": prices
})

# EMA20・EMA50
df["ema20"] = df["close"].ewm(span=20, adjust=False).mean()
df["ema50"] = df["close"].ewm(span=50, adjust=False).mean()

# シグナル
df["signal"] = 0
df.loc[df["ema20"] > df["ema50"], "signal"] = 1
df.loc[df["ema20"] < df["ema50"], "signal"] = -1

# 売買サイン
df["buy_signal"] = (df["signal"].shift(1) == -1) & (df["signal"] == 1)
df["sell_signal"] = (df["signal"].shift(1) == 1) & (df["signal"] == -1)

print("\n--- EMAクロス戦略データ ---")
display(df[["close", "ema20", "ema50", "buy_signal", "sell_signal"]].tail(30))

# グラフ表示
plt.figure(figsize=(14, 6))
plt.plot(df["close"], label="Close")
plt.plot(df["ema20"], label="EMA20")
plt.plot(df["ema50"], label="EMA50")

buy_points = df.index[df["buy_signal"]]
sell_points = df.index[df["sell_signal"]]

plt.scatter(buy_points, df.loc[buy_points, "close"], marker="^", label="Buy Signal")
plt.scatter(sell_points, df.loc[sell_points, "close"], marker="v", label="Sell Signal")

plt.title("BTC/JPY EMA Cross Strategy")
plt.xlabel("Index")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.show()

# 超シンプルな売買結果
cash = 1000000  # 100万円スタート
position = 0
trades = []

for i in range(len(df)):
    price = df.loc[i, "close"]

    if df.loc[i, "buy_signal"] and position == 0:
        position = cash / price
        cash = 0
        trades.append({"type": "BUY", "price": price, "index": i})

    elif df.loc[i, "sell_signal"] and position > 0:
        cash = position * price
        position = 0
        trades.append({"type": "SELL", "price": price, "index": i})

final_value = cash if position == 0 else position * df.iloc[-1]["close"]

print("\n=========================================")
print("3週目の簡易結果")
print("=========================================")
print("最終資産:", round(final_value, 0), "円")
print("売買回数:", len(trades))
print("最初の売買履歴:", trades[:10])



# =========================================
# 4週目 完全版
# EMAクロス戦略をちゃんと検証する
# bitFlyer / BTC/JPY 前提の練習用
# =========================================

!pip -q install pandas matplotlib numpy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================================
# 1. 練習用価格データを作る
# =========================================
np.random.seed(42)

rows = 500
base_price = 10000000  # BTC/JPYのイメージ

price_changes = np.random.normal(loc=3000, scale=50000, size=rows)
prices = base_price + np.cumsum(price_changes)

df = pd.DataFrame({
    "close": prices
})

# =========================================
# 2. EMA20 / EMA50
# =========================================
df["ema20"] = df["close"].ewm(span=20, adjust=False).mean()
df["ema50"] = df["close"].ewm(span=50, adjust=False).mean()

# =========================================
# 3. 売買サイン
# =========================================
df["signal"] = 0
df.loc[df["ema20"] > df["ema50"], "signal"] = 1
df.loc[df["ema20"] < df["ema50"], "signal"] = -1

df["buy_signal"] = (df["signal"].shift(1) == -1) & (df["signal"] == 1)
df["sell_signal"] = (df["signal"].shift(1) == 1) & (df["signal"] == -1)

# =========================================
# 4. バックテスト設定
# =========================================
initial_capital = 1_000_000   # 100万円
fee_rate = 0.              # 0.1%
slippage = 0.             # 0.05%

cash = initial_capital
position = 0.0
entry_price = 0.0

equity_curve = []
trade_pairs = []
current_trade = None

# =========================================
# 5. バックテスト本体
# =========================================
for i in range(len(df)):
    price = df.loc[i, "close"]

    # 買い
    if df.loc[i, "buy_signal"] and position == 0:
        buy_price = price * (1 + slippage)
        fee = cash * fee_rate
        qty = (cash - fee) / buy_price

        if qty > 0:
            position = qty
            entry_price = buy_price
            cash = 0.0

            current_trade = {
                "buy_index": i,
                "buy_price": buy_price
            }

    # 売り
    elif df.loc[i, "sell_signal"] and position > 0:
        sell_price = price * (1 - slippage)
        proceeds = position * sell_price
        fee = proceeds * fee_rate
        cash = proceeds - fee

        if current_trade is not None:
            return_pct = (sell_price / current_trade["buy_price"] - 1) * 100
            pnl = cash - initial_capital if len(trade_pairs) == 0 else None

            trade_pairs.append({
                "buy_index": current_trade["buy_index"],
                "sell_index": i,
                "buy_price": current_trade["buy_price"],
                "sell_price": sell_price,
                "return_pct": return_pct
            })

        position = 0.0
        entry_price = 0.0
        current_trade = None

    # 現在資産
    equity = cash + position * price
    equity_curve.append(equity)

df["equity"] = equity_curve

# 未決済なら最後の価格で評価だけする
final_equity = df["equity"].iloc[-1]

# =========================================
# 6. 取引履歴
# =========================================
trades_df = pd.DataFrame(trade_pairs)

# =========================================
# 7. 指標計算
# =========================================
total_return = (final_equity / initial_capital - 1) * 100

df["peak"] = df["equity"].cummax()
df["drawdown"] = (df["equity"] / df["peak"] - 1) * 100
max_drawdown = df["drawdown"].min()

num_trades = len(trades_df)

if num_trades > 0:
    win_rate = (trades_df["return_pct"] > 0).mean() * 100
    avg_return = trades_df["return_pct"].mean()
    best_trade = trades_df["return_pct"].max()
    worst_trade = trades_df["return_pct"].min()
else:
    win_rate = 0.0
    avg_return = 0.0
    best_trade = 0.0
    worst_trade = 0.0

# =========================================
# 8. 結果表示
# =========================================
print("=" * 60)
print("4週目：EMAクロス戦略 バックテスト結果")
print("=" * 60)
print(f"初期資金: {initial_capital:,.0f} 円")
print(f"最終資産: {final_equity:,.0f} 円")
print(f"総リターン: {total_return:.2f} %")
print(f"売買回数: {num_trades}")
print(f"勝率: {win_rate:.2f} %")
print(f"平均トレード損益率: {avg_return:.2f} %")
print(f"最大ドローダウン: {max_drawdown:.2f} %")
print(f"最大勝ちトレード: {best_trade:.2f} %")
print(f"最大負けトレード: {worst_trade:.2f} %")
print("=" * 60)

print("\n--- 取引履歴（直近10件）---")
display(trades_df.tail(10) if len(trades_df) > 0 else pd.DataFrame())

# =========================================
# 9. グラフ：価格と売買ポイント
# =========================================
plt.figure(figsize=(14, 6))
plt.plot(df["close"], label="Close")
plt.plot(df["ema20"], label="EMA20")
plt.plot(df["ema50"], label="EMA50")

buy_points = df.index[df["buy_signal"]]
sell_points = df.index[df["sell_signal"]]

plt.scatter(buy_points, df.loc[buy_points, "close"], marker="^", label="Buy")
plt.scatter(sell_points, df.loc[sell_points, "close"], marker="v", label="Sell")

plt.title("BTC/JPY EMA Cross")
plt.xlabel("Index")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.show()

# =========================================
# 10. グラフ：資産推移
# =========================================
plt.figure(figsize=(14, 5))
plt.plot(df["equity"], label="Equity Curve")
plt.title("資産推移")
plt.xlabel("Index")
plt.ylabel("Equity")
plt.legend()
plt.grid(True)
plt.show()

# =========================================
# 11. グラフ：ドローダウン
# =========================================
plt.figure(figsize=(14, 4))
plt.fill_between(df.index, df["drawdown"], 0, alpha=0.3)
plt.title("ドローダウン (%)")
plt.xlabel("Index")
plt.ylabel("Drawdown %")
plt.grid(True)
plt.show()


# =========================================
# 5週目 完全版
# EMAクロス + ATR損切り
# bitFlyer / BTC/JPY 前提の練習用
# =========================================

!pip -q install pandas matplotlib numpy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================================
# 1. 練習用OHLCデータを作る
# =========================================
np.random.seed(42)

rows = 500
base_price = 10_000_000

price_changes = np.random.normal(loc=3000, scale=50000, size=rows)
close = base_price + np.cumsum(price_changes)

open_price = close + np.random.normal(0, 20000, rows)
high = np.maximum(open_price, close) + np.random.uniform(10000, 60000, rows)
low = np.minimum(open_price, close) - np.random.uniform(10000, 60000, rows)

df = pd.DataFrame({
    "open": open_price,
    "high": high,
    "low": low,
    "close": close
})

# =========================================
# 2. EMA20 / EMA50
# =========================================
df["ema20"] = df["close"].ewm(span=20, adjust=False).mean()
df["ema50"] = df["close"].ewm(span=50, adjust=False).mean()

# =========================================
# 3. ATRを計算
# =========================================
df["prev_close"] = df["close"].shift(1)

df["tr1"] = df["high"] - df["low"]
df["tr2"] = (df["high"] - df["prev_close"]).abs()
df["tr3"] = (df["low"] - df["prev_close"]).abs()

df["tr"] = df[["tr1", "tr2", "tr3"]].max(axis=1)
df["atr"] = df["tr"].rolling(14).mean()

# =========================================
# 4. 売買サイン
# =========================================
df["signal"] = 0
df.loc[df["ema20"] > df["ema50"], "signal"] = 1
df.loc[df["ema20"] < df["ema50"], "signal"] = -1

df["buy_signal"] = (df["signal"].shift(1) == -1) & (df["signal"] == 1)
df["sell_signal"] = (df["signal"].shift(1) == 1) & (df["signal"] == -1)

# =========================================
# 5. バックテスト設定
# =========================================
initial_capital = 1_000_000
fee_rate = 0.001
slippage = 0.0005

atr_stop_multiplier = 2.0  # ATRの2倍で損切り

cash = initial_capital
position = 0.0
entry_price = 0.0
stop_price = 0.0

equity_curve = []
trade_pairs = []
current_trade = None

# =========================================
# 6. バックテスト本体
# =========================================
for i in range(len(df)):
    row = df.iloc[i]

    price = row["close"]
    high_price = row["high"]
    low_price = row["low"]
    atr = row["atr"]

    # 現在資産
    equity = cash + position * price
    equity_curve.append(equity)

    # ATRがまだ計算できない期間はスキップ
    if np.isnan(atr):
        continue

    # =========================
    # 損切り判定
    # =========================
    if position > 0 and low_price <= stop_price:
        sell_price = stop_price * (1 - slippage)
        proceeds = position * sell_price
        fee = proceeds * fee_rate
        cash = proceeds - fee

        if current_trade is not None:
            return_pct = (sell_price / current_trade["buy_price"] - 1) * 100

            trade_pairs.append({
                "buy_index": current_trade["buy_index"],
                "sell_index": i,
                "buy_price": current_trade["buy_price"],
                "sell_price": sell_price,
                "return_pct": return_pct,
                "reason": "ATR_STOP"
            })

        position = 0.0
        entry_price = 0.0
        stop_price = 0.0
        current_trade = None
        continue

    # =========================
    # EMA売り判定
    # =========================
    if position > 0 and row["sell_signal"]:
        sell_price = price * (1 - slippage)
        proceeds = position * sell_price
        fee = proceeds * fee_rate
        cash = proceeds - fee

        if current_trade is not None:
            return_pct = (sell_price / current_trade["buy_price"] - 1) * 100

            trade_pairs.append({
                "buy_index": current_trade["buy_index"],
                "sell_index": i,
                "buy_price": current_trade["buy_price"],
                "sell_price": sell_price,
                "return_pct": return_pct,
                "reason": "EMA_SELL"
            })

        position = 0.0
        entry_price = 0.0
        stop_price = 0.0
        current_trade = None
        continue

    # =========================
    # 買い判定
    # =========================
    if position == 0 and row["buy_signal"]:
        buy_price = price * (1 + slippage)
        fee = cash * fee_rate
        qty = (cash - fee) / buy_price

        if qty > 0:
            position = qty
            entry_price = buy_price
            cash = 0.0

            # ATR損切り価格
            stop_price = entry_price - atr * atr_stop_multiplier

            current_trade = {
                "buy_index": i,
                "buy_price": buy_price,
                "stop_price": stop_price
            }

df["equity"] = equity_curve

final_equity = df["equity"].iloc[-1]
trades_df = pd.DataFrame(trade_pairs)

# =========================================
# 7. 指標計算
# =========================================
total_return = (final_equity / initial_capital - 1) * 100

df["peak"] = df["equity"].cummax()
df["drawdown"] = (df["equity"] / df["peak"] - 1) * 100
max_drawdown = df["drawdown"].min()

num_trades = len(trades_df)

if num_trades > 0:
    win_rate = (trades_df["return_pct"] > 0).mean() * 100
    avg_return = trades_df["return_pct"].mean()
    best_trade = trades_df["return_pct"].max()
    worst_trade = trades_df["return_pct"].min()
    stop_count = (trades_df["reason"] == "ATR_STOP").sum()
else:
    win_rate = 0
    avg_return = 0
    best_trade = 0
    worst_trade = 0
    stop_count = 0

# =========================================
# 8. 結果表示
# =========================================
print("=" * 60)
print("5週目：EMAクロス + ATR損切り バックテスト結果")
print("=" * 60)
print(f"初期資金: {initial_capital:,.0f} 円")
print(f"最終資産: {final_equity:,.0f} 円")
print(f"総リターン: {total_return:.2f} %")
print(f"売買回数: {num_trades}")
print(f"勝率: {win_rate:.2f} %")
print(f"平均トレード損益率: {avg_return:.2f} %")
print(f"最大ドローダウン: {max_drawdown:.2f} %")
print(f"最大勝ちトレード: {best_trade:.2f} %")
print(f"最大負けトレード: {worst_trade:.2f} %")
print(f"ATR損切り回数: {stop_count}")
print("=" * 60)

print("\n--- 取引履歴（直近10件）---")
display(trades_df.tail(10) if len(trades_df) > 0 else pd.DataFrame())

# =========================================
# 9. グラフ：価格・EMA・売買ポイント
# =========================================
plt.figure(figsize=(14, 6))
plt.plot(df["close"], label="Close")
plt.plot(df["ema20"], label="EMA20")
plt.plot(df["ema50"], label="EMA50")

buy_points = df.index[df["buy_signal"]]
sell_points = df.index[df["sell_signal"]]

plt.scatter(buy_points, df.loc[buy_points, "close"], marker="^", label="Buy")
plt.scatter(sell_points, df.loc[sell_points, "close"], marker="v", label="Sell")

plt.title("BTC/JPY EMA Cross + ATR Stop")
plt.xlabel("Index")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.show()

# =========================================
# 10. グラフ：資産推移
# =========================================
plt.figure(figsize=(14, 5))
plt.plot(df["equity"], label="Equity Curve")
plt.title("資産推移")
plt.xlabel("Index")
plt.ylabel("Equity")
plt.legend()
plt.grid(True)
plt.show()

# =========================================
# 11. グラフ：ドローダウン
# =========================================
plt.figure(figsize=(14, 4))
plt.fill_between(df.index, df["drawdown"], 0, alpha=0.3)
plt.title("ドローダウン (%)")
plt.xlabel("Index")
plt.ylabel("Drawdown %")
plt.grid(True)
plt.show()


# =========================================
# 6週目 完全版
# EMAクロス + ATR損切り + 利確 + トレーリングストップ
# bitFlyer / BTC/JPY 前提の練習用
# =========================================

!pip -q install pandas matplotlib numpy

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================================
# 1. 練習用OHLCデータを作る
# =========================================
np.random.seed(42)

rows = 500
base_price = 10_000_000

price_changes = np.random.normal(loc=3000, scale=50000, size=rows)
close = base_price + np.cumsum(price_changes)

open_price = close + np.random.normal(0, 20000, rows)
high = np.maximum(open_price, close) + np.random.uniform(10000, 60000, rows)
low = np.minimum(open_price, close) - np.random.uniform(10000, 60000, rows)

df = pd.DataFrame({
    "open": open_price,
    "high": high,
    "low": low,
    "close": close
})

# =========================================
# 2. EMA20 / EMA50
# =========================================
df["ema20"] = df["close"].ewm(span=20, adjust=False).mean()
df["ema50"] = df["close"].ewm(span=50, adjust=False).mean()

# =========================================
# 3. ATR
# =========================================
df["prev_close"] = df["close"].shift(1)

df["tr1"] = df["high"] - df["low"]
df["tr2"] = (df["high"] - df["prev_close"]).abs()
df["tr3"] = (df["low"] - df["prev_close"]).abs()

df["tr"] = df[["tr1", "tr2", "tr3"]].max(axis=1)
df["atr"] = df["tr"].rolling(14).mean()

# =========================================
# 4. EMAクロス売買サイン
# =========================================
df["signal"] = 0
df.loc[df["ema20"] > df["ema50"], "signal"] = 1
df.loc[df["ema20"] < df["ema50"], "signal"] = -1

df["buy_signal"] = (df["signal"].shift(1) == -1) & (df["signal"] == 1)
df["sell_signal"] = (df["signal"].shift(1) == 1) & (df["signal"] == -1)

# =========================================
# 5. バックテスト設定
# =========================================
initial_capital = 1_000_000
fee_rate = 0.001
slippage = 0.0005

atr_stop_multiplier = 2.0      # 損切り：ATRの2倍
take_profit_multiplier = 3.0   # 利確：ATRの3倍
trailing_multiplier = 2.0      # トレーリング：最高値からATRの2倍

cash = initial_capital
position = 0.0
entry_price = 0.0

stop_price = 0.0
take_profit_price = 0.0
trailing_stop_price = 0.0
highest_price = 0.0

equity_curve = []
trade_pairs = []
current_trade = None

# =========================================
# 6. バックテスト本体
# =========================================
for i in range(len(df)):
    row = df.iloc[i]

    price = row["close"]
    high_price = row["high"]
    low_price = row["low"]
    atr = row["atr"]

    # 現在資産
    equity = cash + position * price
    equity_curve.append(equity)

    # ATRがまだない期間はスキップ
    if np.isnan(atr):
        continue

    # =====================================
    # 保有中の処理
    # =====================================
    if position > 0:
        # 最高値を更新
        highest_price = max(highest_price, high_price)

        # トレーリングストップ更新
        trailing_stop_price = highest_price - atr * trailing_multiplier

        exit_reason = None
        sell_price = None

        # 1. ATR損切り
        if low_price <= stop_price:
            exit_reason = "ATR_STOP"
            sell_price = stop_price * (1 - slippage)

        # 2. 利確
        elif high_price >= take_profit_price:
            exit_reason = "TAKE_PROFIT"
            sell_price = take_profit_price * (1 - slippage)

        # 3. トレーリングストップ
        elif low_price <= trailing_stop_price:
            exit_reason = "TRAILING_STOP"
            sell_price = trailing_stop_price * (1 - slippage)

        # 4. EMA売り
        elif row["sell_signal"]:
            exit_reason = "EMA_SELL"
            sell_price = price * (1 - slippage)

        # 決済処理
        if exit_reason is not None:
            proceeds = position * sell_price
            fee = proceeds * fee_rate
            cash = proceeds - fee

            if current_trade is not None:
                return_pct = (sell_price / current_trade["buy_price"] - 1) * 100

                trade_pairs.append({
                    "buy_index": current_trade["buy_index"],
                    "sell_index": i,
                    "buy_price": current_trade["buy_price"],
                    "sell_price": sell_price,
                    "return_pct": return_pct,
                    "reason": exit_reason
                })

            position = 0.0
            entry_price = 0.0
            stop_price = 0.0
            take_profit_price = 0.0
            trailing_stop_price = 0.0
            highest_price = 0.0
            current_trade = None

            continue

    # =====================================
    # 新規買い
    # =====================================
    if position == 0 and row["buy_signal"]:
        buy_price = price * (1 + slippage)
        fee = cash * fee_rate
        qty = (cash - fee) / buy_price

        if qty > 0:
            position = qty
            entry_price = buy_price
            cash = 0.0

            stop_price = entry_price - atr * atr_stop_multiplier
            take_profit_price = entry_price + atr * take_profit_multiplier
            highest_price = entry_price
            trailing_stop_price = entry_price - atr * trailing_multiplier

            current_trade = {
                "buy_index": i,
                "buy_price": buy_price,
                "stop_price": stop_price,
                "take_profit_price": take_profit_price,
                "trailing_stop_price": trailing_stop_price
            }

df["equity"] = equity_curve

final_equity = df["equity"].iloc[-1]
trades_df = pd.DataFrame(trade_pairs)

# =========================================
# 7. 指標計算
# =========================================
total_return = (final_equity / initial_capital - 1) * 100

df["peak"] = df["equity"].cummax()
df["drawdown"] = (df["equity"] / df["peak"] - 1) * 100
max_drawdown = df["drawdown"].min()

num_trades = len(trades_df)

if num_trades > 0:
    win_rate = (trades_df["return_pct"] > 0).mean() * 100
    avg_return = trades_df["return_pct"].mean()
    best_trade = trades_df["return_pct"].max()
    worst_trade = trades_df["return_pct"].min()

    stop_count = (trades_df["reason"] == "ATR_STOP").sum()
    tp_count = (trades_df["reason"] == "TAKE_PROFIT").sum()
    trail_count = (trades_df["reason"] == "TRAILING_STOP").sum()
    ema_sell_count = (trades_df["reason"] == "EMA_SELL").sum()
else:
    win_rate = 0
    avg_return = 0
    best_trade = 0
    worst_trade = 0

    stop_count = 0
    tp_count = 0
    trail_count = 0
    ema_sell_count = 0

# =========================================
# 8. 結果表示
# =========================================
print("=" * 70)
print("6週目：EMAクロス + ATR損切り + 利確 + トレーリング")
print("=" * 70)
print(f"初期資金: {initial_capital:,.0f} 円")
print(f"最終資産: {final_equity:,.0f} 円")
print(f"総リターン: {total_return:.2f} %")
print(f"売買回数: {num_trades}")
print(f"勝率: {win_rate:.2f} %")
print(f"平均トレード損益率: {avg_return:.2f} %")
print(f"最大ドローダウン: {max_drawdown:.2f} %")
print(f"最大勝ちトレード: {best_trade:.2f} %")
print(f"最大負けトレード: {worst_trade:.2f} %")
print("-" * 70)
print(f"ATR損切り回数: {stop_count}")
print(f"利確回数: {tp_count}")
print(f"トレーリング決済回数: {trail_count}")
print(f"EMA売り回数: {ema_sell_count}")
print("=" * 70)

print("\n--- 取引履歴（直近10件）---")
display(trades_df.tail(10) if len(trades_df) > 0 else pd.DataFrame())

# =========================================
# 9. グラフ：価格・EMA・売買ポイント
# =========================================
plt.figure(figsize=(14, 6))
plt.plot(df["close"], label="Close")
plt.plot(df["ema20"], label="EMA20")
plt.plot(df["ema50"], label="EMA50")

buy_points = df.index[df["buy_signal"]]
sell_points = df.index[df["sell_signal"]]

plt.scatter(buy_points, df.loc[buy_points, "close"], marker="^", label="Buy Signal")
plt.scatter(sell_points, df.loc[sell_points, "close"], marker="v", label="EMA Sell Signal")

plt.title("BTC/JPY EMA Cross + ATR Stop + Take Profit + Trailing")
plt.xlabel("Index")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.show()

# =========================================
# 10. グラフ：資産推移
# =========================================
plt.figure(figsize=(14, 5))
plt.plot(df["equity"], label="Equity Curve")
plt.title("資産推移")
plt.xlabel("Index")
plt.ylabel("Equity")
plt.legend()
plt.grid(True)
plt.show()

# =========================================
# 11. グラフ：ドローダウン
# =========================================
plt.figure(figsize=(14, 4))
plt.fill_between(df.index, df["drawdown"], 0, alpha=0.3)
plt.title("ドローダウン (%)")
plt.xlabel("Index")
plt.ylabel("Drawdown %")
plt.grid(True)
plt.show()

# =========================================
# 12. 決済理由の集計グラフ
# =========================================
if len(trades_df) > 0:
    reason_counts = trades_df["reason"].value_counts()

    plt.figure(figsize=(8, 4))
    plt.bar(reason_counts.index, reason_counts.values)
    plt.title("決済理由の回数")
    plt.xlabel("Reason")
    plt.ylabel("Count")
    plt.xticks(rotation=30)
    plt.grid(True, axis="y")
    plt.show()

import pandas as pd

df = pd.read_csv("bitflyer_btcjpy_1h.csv")
df["datetime"] = pd.to_datetime(df["datetime"])

print(df.head())
print(df.tail())
print("件数:", len(df))

print(df.columns)

# =========================================
# bitFlyer BTC/JPY 1時間足CSV専用
# EMAクロス + ATR損切り + 利確 + トレーリング
# =========================================

!pip -q install pandas numpy matplotlib

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================================
# 1. CSV読み込み
# =========================================
CSV_FILE = "bitflyer_btcjpy_1h.csv"

df = pd.read_csv(CSV_FILE)
df["datetime"] = pd.to_datetime(df["datetime"])

df = df.sort_values("datetime").reset_index(drop=True)

print("読み込み完了")
print("件数:", len(df))
print("最古:", df["datetime"].min())
print("最新:", df["datetime"].max())
display(df.head())
display(df.tail())

# =========================================
# 2. 必要列チェック
# =========================================
required_cols = ["datetime", "open", "high", "low", "close", "volume"]

missing = [c for c in required_cols if c not in df.columns]

if missing:
    raise ValueError(f"必要な列がありません: {missing}")

# =========================================
# 3. パラメータ設定
# =========================================
initial_capital = 1_000_000   # まずは100万円で検証
fee_rate = 0.001              # 手数料 0.1%
slippage = 0.0005             # スリッページ 0.05%

ema_fast = 8
ema_slow = 21

atr_stop_multiplier = 1.3
take_profit_multiplier = 2.8
trailing_multiplier = 2.5
atr_period = 14 # ATRの期間を追加

# =========================================
# 4. 指標計算
# =========================================
df["ema20"] = df["close"].ewm(span=ema_fast, adjust=False).mean()
df["ema50"] = df["close"].ewm(span=ema_slow, adjust=False).mean()

# ATR計算
df["prev_close"] = df["close"].shift(1)
df["tr1"] = df["high"] - df["low"]
df["tr2"] = (df["high"] - df["prev_close"]).abs()
df["tr3"] = (df["low"] - df["prev_close"]).abs()
df["tr"] = df[["tr1", "tr2", "tr3"]].max(axis=1)
df["atr"] = df["tr"].rolling(atr_period).mean()

# =========================================
# 5. 売買シグナル
# =========================================
df["signal"] = 0
df.loc[df["ema20"] > df["ema50"], "signal"] = 1
df.loc[df["ema20"] < df["ema50"], "signal"] = -1

df["buy_signal"] = (df["signal"].shift(1) == -1) & (df["signal"] == 1)
df["sell_signal"] = (df["signal"].shift(1) == 1) & (df["signal"] == -1)

# =========================================
# 6. バックテスト本体
# =========================================
cash = initial_capital
position = 0.0
entry_price = 0.0

stop_price = 0.0
take_profit_price = 0.0
trailing_stop_price = 0.0
highest_price = 0.0

equity_curve = []
trade_pairs = []
current_trade = None

for i in range(len(df)):
    row = df.iloc[i]

    price = row["close"]
    high_price = row["high"]
    low_price = row["low"]
    atr = row["atr"]

    # 現在資産
    equity = cash + position * price
    equity_curve.append(equity)

    # ATRがない期間はスキップ
    if np.isnan(atr):
        continue

    # =========================
    # 保有中の決済判定
    # =========================
    if position > 0:
        highest_price = max(highest_price, high_price)
        trailing_stop_price = highest_price - atr * trailing_multiplier

        exit_reason = None
        sell_price = None

        # 1. ATR損切り
        if low_price <= stop_price:
            exit_reason = "ATR_STOP"
            sell_price = stop_price * (1 - slippage)

        # 2. 利確
        elif high_price >= take_profit_price:
            exit_reason = "TAKE_PROFIT"
            sell_price = take_profit_price * (1 - slippage)

        # 3. トレーリングストップ
        elif low_price <= trailing_stop_price:
            exit_reason = "TRAILING_STOP"
            sell_price = trailing_stop_price * (1 - slippage)

        # 4. EMA売り
        elif row["sell_signal"]:
            exit_reason = "EMA_SELL"
            sell_price = price * (1 - slippage)

        if exit_reason is not None:
            proceeds = position * sell_price
            fee = proceeds * fee_rate
            cash = proceeds - fee

            if current_trade is not None:
                return_pct = (sell_price / current_trade["buy_price"] - 1) * 100
                pnl = cash - current_trade["cash_before"]

                trade_pairs.append({
                    "buy_datetime": current_trade["buy_datetime"],
                    "sell_datetime": row["datetime"],
                    "buy_index": current_trade["buy_index"],
                    "sell_index": i,
                    "buy_price": current_trade["buy_price"],
                    "sell_price": sell_price,
                    "return_pct": return_pct,
                    "pnl": pnl,
                    "reason": exit_reason
                })

            position = 0.0
            entry_price = 0.0
            stop_price = 0.0
            take_profit_price = 0.0
            trailing_stop_price = 0.0
            highest_price = 0.0
            current_trade = None

            continue

    # =========================
    # 新規買い判定
    # =========================
    if position == 0 and row["buy_signal"]:
        buy_price = price * (1 + slippage)
        fee = cash * fee_rate
        qty = (cash - fee) / buy_price

        if qty > 0:
            cash_before = cash

            position = qty
            entry_price = buy_price
            cash = 0.0

            stop_price = entry_price - atr * atr_stop_multiplier
            take_profit_price = entry_price + atr * take_profit_multiplier
            highest_price = entry_price
            trailing_stop_price = entry_price - atr * trailing_multiplier

            current_trade = {
                "buy_datetime": row["datetime"],
                "buy_index": i,
                "buy_price": buy_price,
                "cash_before": cash_before,
                "stop_price": stop_price,
                "take_profit_price": take_profit_price,
                "trailing_stop_price": trailing_stop_price
            }

df["equity"] = equity_curve

final_equity = df["equity"].iloc[-1]
trades_df = pd.DataFrame(trade_pairs)

# =========================================
# 7. 成績計算
# =========================================
total_return = (final_equity / initial_capital - 1) * 100

df["peak"] = df["equity"].cummax()
df["drawdown"] = (df["equity"] / df["peak"] - 1) * 100
max_drawdown = df["drawdown"].min()

num_trades = len(trades_df)

if num_trades > 0:
    win_rate = (trades_df["pnl"] > 0).mean() * 100
    avg_return = trades_df["return_pct"].mean()
    best_trade = trades_df["return_pct"].max()
    worst_trade = trades_df["return_pct"].min()

    gross_profit = trades_df.loc[trades_df["pnl"] > 0, "pnl"].sum()
    gross_loss = abs(trades_df.loc[trades_df["pnl"] < 0, "pnl"].sum())
    profit_factor = gross_profit / gross_loss if gross_loss > 0 else np.nan

    stop_count = (trades_df["reason"] == "ATR_STOP").sum()
    tp_count = (trades_df["reason"] == "TAKE_PROFIT").sum()
    trail_count = (trades_df["reason"] == "TRAILING_STOP").sum()
    ema_sell_count = (trades_df["reason"] == "EMA_SELL").sum()
else:
    win_rate = 0
    avg_return = 0
    best_trade = 0
    worst_trade = 0
    profit_factor = np.nan

    stop_count = 0
    tp_count = 0
    trail_count = 0
    ema_sell_count = 0

# =========================================
# 8. 月次成績
# =========================================
df["month"] = df["datetime"].dt.to_period("M").astype(str)

monthly = df.groupby("month").agg(
    first_equity=("equity", "first"),
    last_equity=("equity", "last")
).reset_index()

monthly["monthly_return_pct"] = (
    monthly["last_equity"] / monthly["first_equity"] - 1
) * 100

# =========================================
# 9. 結果表示
# =========================================
print("=" * 70)
print("bitFlyer BTC/JPY CSVバックテスト結果")
print("=" * 70)
print(f"初期資金: {initial_capital:,.0f} 円")
print(f"最終資産: {final_equity:,.0f} 円")
print(f"総リターン: {total_return:.2f} %")
print(f"最大ドローダウン: {max_drawdown:.2f} %")
print(f"売買回数: {num_trades}")
print(f"勝率: {win_rate:.2f} %")
print(f"平均トレード損益率: {avg_return:.2f} %")
print(f"Profit Factor: {profit_factor:.2f}")
print("-" * 70)
print(f"ATR損切り回数: {stop_count}")
print(f"利確回数: {tp_count}")
print(f"トレーリング決済回数: {trail_count}")
print(f"EMA売り回数: {ema_sell_count}")
print("=" * 70)

print("\n--- 取引履歴（直近10件）---")
display(trades_df.tail(10) if len(trades_df) > 0 else pd.DataFrame())

print("\n--- 月次成績（直近24ヶ月）---")
display(monthly.tail(24))

# =========================================
# 10. グラフ：価格とEMA
# =========================================
plt.figure(figsize=(14, 6))
plt.plot(df["datetime"], df["close"], label="Close")
plt.plot(df["datetime"], df["ema20"], label="EMA20")
plt.plot(df["datetime"], df["ema50"], label="EMA50")
plt.title("BTC/JPY Price + EMA")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.show()

# =========================================
# 11. グラフ：資産推移
# =========================================
plt.figure(figsize=(14, 5))
plt.plot(df["datetime"], df["equity"], label="Equity Curve")
plt.title("資産推移")
plt.xlabel("Date")
plt.ylabel("Equity")
plt.legend()
plt.grid(True)
plt.show()

# =========================================
# 12. グラフ：ドローダウン
# =========================================
plt.figure(figsize=(14, 4))
plt.fill_between(df["datetime"], df["drawdown"], 0, alpha=0.3)
plt.title("ドローダウン (%)")
plt.xlabel("Date")
plt.ylabel("Drawdown %")
plt.grid(True)
plt.show()

# =========================================
# 13. 決済理由
# =========================================
if len(trades_df) > 0:
    reason_counts = trades_df["reason"].value_counts()

    plt.figure(figsize=(8, 4))
    plt.bar(reason_counts.index, reason_counts.values)
    plt.title("決済理由の回数")
    plt.xlabel("Reason")
    plt.ylabel("Count")
    plt.xticks(rotation=30)
    plt.grid(True, axis="y")
    plt.show()

# =========================================
# 14. 結果をCSV保存
# =========================================
trades_df.to_csv("backtest_trades_result.csv", index=False, encoding="utf-8-sig")
monthly.to_csv("backtest_monthly_result.csv", index=False, encoding="utf-8-sig")

print("保存完了:")
print("backtest_trades_result.csv")
print("backtest_monthly_result.csv")
```

```python
# =========================================
# bitFlyer BTC/JPY CSV専用
# トレンド＋押し目戦略
# =========================================

!pip -q install pandas numpy matplotlib

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================
# 1. CSV読み込み
# =========================
CSV_FILE = "bitflyer_btcjpy_1h.csv"

df = pd.read_csv(CSV_FILE)
df["datetime"] = pd.to_datetime(df["datetime"])
df = df.sort_values("datetime").reset_index(drop=True)


print("件数:", len(df))
print("最古:", df["datetime"].min())
print("最新:", df["datetime"].max())

# =========================
# 2. パラメータ
# =========================
initial_capital = 1_000_000

fee_rate = 0.001
slippage = 0.0005

ema_fast = 20
ema_slow = 50
atr_period = 14

pullback_range = 0.002
atr_stop_multiplier = 1.2
take_profit_multiplier = 3.2

# =========================
# 3. 指標計算
# =========================
df["ema20"] = df["close"].ewm(span=ema_fast, adjust=False).mean()
df["ema50"] = df["close"].ewm(span=ema_slow, adjust=False).mean()

df["prev_close"] = df["close"].shift(1)
df["tr1"] = df["high"] - df["low"]
df["tr2"] = (df["high"] - df["prev_close"]).abs()
df["tr3"] = (df["low"] - df["prev_close"]).abs()
df["tr"] = df[["tr1", "tr2", "tr3"]].max(axis=1)
df["atr"] = df["tr"].rolling(atr_period).mean()

# =========================
# 4. エントリー条件
# =========================
df["trend_up"] = df["ema20"] > df["ema50"]

df["near_ema20"] = abs(df["close"] - df["ema20"]) / df["close"] < pullback_range

df["bullish"] = df["close"] > df["open"]

df["hour"] = df["datetime"].dt.hour

df["time_filter"] = (df["hour"] >= 9) & (df["hour"] <= 23)

df["trend_strength"] = abs(df["ema20"] - df["ema50"]) / df["close"]

df["buy_signal"] = (
    df["trend_up"] &
    df["near_ema20"] &
    df["bullish"] &
    df["time_filter"] &
    (df["trend_strength"] > 0.002)
)

# =========================
# 5. バックテスト
# =========================
cash = initial_capital
position = 0.0
entry_price = 0.0

stop_price = 0.0
take_profit_price = 0.0

equity_curve = []
trades = []
current_trade = None

for i in range(len(df)):
    row = df.iloc[i]

    price = row["close"]
    high_price = row["high"]
    low_price = row["low"]
    atr = row["atr"]

    equity = cash + position * price
    equity_curve.append(equity)

    if np.isnan(atr):
        continue

    # 保有中
    if position > 0:
        exit_reason = None
        sell_price = None

        # 損切り
        if low_price <= stop_price:
            exit_reason = "ATR_STOP"
            sell_price = stop_price * (1 - slippage)

        # 利確
        elif high_price >= take_profit_price:
            exit_reason = "TAKE_PROFIT"
            sell_price = take_profit_price * (1 - slippage)

        # トレンド崩れ
        elif row["ema20"] < row["ema50"]:
            exit_reason = "TREND_EXIT"
            sell_price = price * (1 - slippage)

        if exit_reason is not None:
            proceeds = position * sell_price
            fee = proceeds * fee_rate
            cash = proceeds - fee

            return_pct = (sell_price / current_trade["buy_price"] - 1) * 100
            pnl = cash - current_trade["cash_before"]

            trades.append({
                "buy_datetime": current_trade["buy_datetime"],
                "sell_datetime": row["datetime"],
                "buy_price": current_trade["buy_price"],
                "sell_price": sell_price,
                "return_pct": return_pct,
                "pnl": pnl,
                "reason": exit_reason
            })

            position = 0.0
            entry_price = 0.0
            stop_price = 0.0
            take_profit_price = 0.0
            current_trade = None

            continue

    # 新規買い
    if position == 0 and row["buy_signal"]:
        buy_price = price * (1 + slippage)
        fee = cash * fee_rate
        qty = (cash - fee) / buy_price

        if qty > 0:
            cash_before = cash
            position = qty
            entry_price = buy_price
            cash = 0.0

            stop_price = entry_price - atr * atr_stop_multiplier
            take_profit_price = entry_price + atr * take_profit_multiplier

            current_trade = {
                "buy_datetime": row["datetime"],
                "buy_price": buy_price,
                "cash_before": cash_before
            }

df["equity"] = equity_curve
trades_df = pd.DataFrame(trades)

# =========================
# 6. 成績計算
# =========================
final_equity = df["equity"].iloc[-1]
total_return = (final_equity / initial_capital - 1) * 100

df["peak"] = df["equity"].cummax()
df["drawdown"] = (df["equity"] / df["peak"] - 1) * 100
max_drawdown = df["drawdown"].min()

num_trades = len(trades_df)

if num_trades > 0:
    win_rate = (trades_df["pnl"] > 0).mean() * 100
    avg_return = trades_df["return_pct"].mean()

    gross_profit = trades_df.loc[trades_df["pnl"] > 0, "pnl"].sum()
    gross_loss = abs(trades_df.loc[trades_df["pnl"] < 0, "pnl"].sum())
    profit_factor = gross_profit / gross_loss if gross_loss > 0 else np.nan

    stop_count = (trades_df["reason"] == "ATR_STOP").sum()
    tp_count = (trades_df["reason"] == "TAKE_PROFIT").sum()
    trend_exit_count = (trades_df["reason"] == "TREND_EXIT").sum()
else:
    win_rate = 0
    avg_return = 0
    profit_factor = np.nan
    stop_count = 0
    tp_count = 0
    trend_exit_count = 0

print("=" * 60)
print("トレンド＋押し目戦略 結果")
print("=" * 60)
print(f"初期資金: {initial_capital:,.0f} 円")
print(f"最終資産: {final_equity:,.0f} 円")
print(f"総リターン: {total_return:.2f} %")
print(f"最大ドローダウン: {max_drawdown:.2f} %")
print(f"売買回数: {num_trades}")
print(f"勝率: {win_rate:.2f} %")
print(f"平均トレード損益率: {avg_return:.2f} %")
print(f"Profit Factor: {profit_factor:.2f}")
print("-" * 60)
print(f"ATR損切り回数: {stop_count}")
print(f"利確回数: {tp_count}")
print(f"トレンド終了回数: {trend_exit_count}")
print("=" * 60)

display(trades_df.tail(10) if len(trades_df) > 0 else pd.DataFrame())

# =========================
# 7. グラフ
# =========================
plt.figure(figsize=(14, 6))
plt.plot(df["datetime"], df["close"], label="Close")
plt.plot(df["datetime"], df["ema20"], label="EMA20")
plt.plot(df["datetime"], df["ema50"], label="EMA50")
plt.title("BTC/JPY Trend Pullback Strategy")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(14, 5))
plt.plot(df["datetime"], df["equity"], label="Equity Curve")
plt.title("資産推移")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(14, 4))
plt.fill_between(df["datetime"], df["drawdown"], 0, alpha=0.3)
plt.title("ドローダウン")
plt.grid(True)
plt.show()
```

```python
import time
import hmac
import hashlib
import requests
import json

API_KEY = API_KEY
API_SECRET = API_SECRET

base_url = "https://api.bitflyer.com"
path = "/v1/me/sendchildorder"
method = "POST"

body = json.dumps({
    "product_code": "BTC_JPY",
    "child_order_type": "LIMIT",
    "side": "SELL",
    "price": 13000000,   # ←絶対に約定しない
    "size": 0.001,
    "minute_to_expire": 1
})

timestamp = str(time.time())
text = timestamp + method + path + body

signature = hmac.new(
    API_SECRET.encode("utf-8"),
    text.encode("utf-8"),
    hashlib.sha256
).hexdigest()

headers = {
    "ACCESS-KEY": API_KEY,
    "ACCESS-TIMESTAMP": timestamp,
    "ACCESS-SIGN": signature,
    "Content-Type": "application/json"
}

response = requests.post(base_url + path, headers=headers, data=body)

print("status:", response.status_code)
print(response.text)
```

```python
import requests
import time
import pandas as pd
from datetime import datetime

url = "https://api.bitflyer.com/v1/ticker"

data_list = []

position = 0
entry_price = 0

start_time = time.time()

while True:
    try:
        response = requests.get(url, params={"product_code": "BTC_JPY"})
        data = response.json()

        now = datetime.now()
        price = data["ltp"]

        data_list.append({
            "datetime": now,
            "close": price
        })

        df = pd.DataFrame(data_list)
        df.to_csv("/content/drive/MyDrive/crypto_data/realtime.csv", index=False)

        df["ema20"] = df["close"].ewm(span=20, adjust=False).mean()
        df["ema50"] = df["close"].ewm(span=50, adjust=False).mean()

        signal = "待機"

        if len(df) > 10:
            ema20 = df["ema20"].iloc[-1]
            ema50 = df["ema50"].iloc[-1]
            price_now = df["close"].iloc[-1]

            trend_up = ema20 > ema50
            pullback = abs(price_now - ema20) / price_now < 0.003
            bullish = price_now > df["close"].iloc[-2]
            trend_strength = abs(ema20 - ema50) / price_now
            time_filter = True

            print(f"trend_up={trend_up}, pullback={pullback}, bullish={bullish}, strength={trend_strength:.4f}, time_filter={time_filter}")

            if trend_up and pullback and bullish and trend_strength > 0.0008 and time_filter:
                signal = "買い"

        # =========================
        # 仮想売買のみ：実注文なし
        # =========================
        if signal == "買い" and position == 0:
            position = 1
            entry_price = price

            print("=================================")
            print("▶ 仮想買い")
            print(f"買い価格: {entry_price}")
            print("=================================")

        elif position == 1:
            profit = (price - entry_price) / entry_price * 100

            exit_reason = None

            if profit > 0.3:
                exit_reason = "仮想利確"

            elif profit < -0.2:
                exit_reason = "仮想損切り"

            if exit_reason:
                print("=================================")
                print(f"■ 仮想売り（{exit_reason}）")
                print(f"売り価格: {price}")
                print(f"エントリー: {entry_price}")
                print(f"損益: {profit:.3f}%")
                print("=================================")

                position = 0
                entry_price = 0

        print(
            f"{now.strftime('%H:%M:%S')} | "
            f"価格: {price:.0f} | "
            f"シグナル: {signal} | "
            f"ポジション: {position}"
        )

        time.sleep(60)

        if time.time() - start_time > 60 * 60 * 12:
            print("12時間経過で終了")
            break

    except KeyboardInterrupt:
        print("停止しました")
        break

    except Exception as e:
        print("エラー:", e)
        time.sleep(10)
```
